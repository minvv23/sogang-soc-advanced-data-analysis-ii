```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel('./data/20250601_dataspring_data.xlsx')
```

```python
df['sex'] = df['SQ1'].replace({1:'male', 2:'female'})
df['ko_sex'] = df['SQ1'].replace({1:'남성', 2:'여성'})
df['int_sex'] = df['SQ1'].replace({2:0})
df['anti_immigration'] = 6-df['C5s6']
df['age'] = df['SQ2s2']
df['female_discrimination'] = df['C5s1']
```

```python
model = sm.OLS.from_formula('female_discrimination ~ age', 
                            data=df[df['sex']=='female']).fit(cov_type='HC3')
model.summary()
```

```python
model = sm.OLS.from_formula('female_discrimination ~ age*sex', data=df).fit(cov_type='HC3')
model.summary()
```

```python
fig = sns.lmplot(y='female_discrimination', x='age', hue='sex', data=df)
plt.show()
```
